```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import IterativeImputer
from sklearn.linear_model import BayesianRidge
from sklearn.metrics import mean_absolute_error, r2_score
import numpy as np
import pandas as pd

# Подготовка — берём известные значения
X37_known = X[X["Attr37"].notna()].copy()

np.random.seed(42)
mask = np.random.choice(len(X37_known), size=int(len(X37_known) * 0.2), replace=False)
true_values = X37_known.iloc[mask]["Attr37"].values

results = {}

# 1. Медиана (baseline)
median_val = X37_known["Attr37"].median()
results["Медиана"] = {
    "MAE": mean_absolute_error(true_values, np.full(len(true_values), median_val)),
    "R²":  r2_score(true_values, np.full(len(true_values), median_val))
}

# 2. MICE с BayesianRidge (дефолт)
for name, estimator in [
    ("MICE BayesianRidge", BayesianRidge()),
    ("MICE RandomForest",  RandomForestRegressor(n_estimators=50, random_state=42, n_jobs=-1)),
]:
    X_test = X37_known.copy()
    X_test.iloc[mask, X_test.columns.get_loc("Attr37")] = np.nan
    
    imputer = IterativeImputer(estimator=estimator, max_iter=10, random_state=42, verbose=0)
    X_imp = pd.DataFrame(imputer.fit_transform(X_test), columns=X.columns)
    predicted = X_imp.iloc[mask]["Attr37"].values
    
    results[name] = {
        "MAE": mean_absolute_error(true_values, predicted),
        "R²":  r2_score(true_values, predicted)
    }

# Результаты
print(pd.DataFrame(results).T.round(4))
```

```python
import pandas as pd
import numpy as np
from scipy.io import arff
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.ensemble import RandomForestRegressor

# 1. Загрузка
data, meta = arff.loadarff("1year.arff")
df = pd.DataFrame(data)
X = df.drop(columns=["class"])
y = df["class"]

# 2. Винсоризация Attr37
low  = X["Attr37"].quantile(0.01)
high = X["Attr37"].quantile(0.99)
X["Attr37"] = X["Attr37"].clip(lower=low, upper=high)

# 3. Обучаем импутер только на строках где Attr37 известен
X_known = X[X["Attr37"].notna()].copy()

imputer = IterativeImputer(
    estimator=RandomForestRegressor(n_estimators=50, random_state=42, n_jobs=-1),
    max_iter=10,
    random_state=42,
    verbose=0
)
imputer.fit(X_known)  # обучаем на известных значениях

# 4. Применяем ко всему датасету
X_imputed = pd.DataFrame(imputer.transform(X), columns=X.columns)

# Проверка что пропусков в Attr37 больше нет
print(f"Пропусков в Attr37 до:   {X['Attr37'].isna().sum()}")
print(f"Пропусков в Attr37 после: {X_imputed['Attr37'].isna().sum()}")
print(f"\nПервые 5 заполненных значений:")
print(X_imputed.loc[X["Attr37"].isna(), "Attr37"].head())
```

```python
import os

os.makedirs("imputed_data", exist_ok=True)

for year in ["1year", "2year", "3year", "4year", "5year"]:
    data, _ = arff.loadarff(f"{year}.arff")
    df_year = pd.DataFrame(data)
    X_year = df_year.drop(columns=["class"])
    y_year = df_year["class"]
    
    # Винсоризация
    for col in X_year.columns:
        low  = X_year[col].quantile(0.01)
        high = X_year[col].quantile(0.99)
        X_year[col] = X_year[col].clip(lower=low, upper=high)
    
    # MICE
    imputer = IterativeImputer(
        estimator=RandomForestRegressor(n_estimators=50, random_state=42, n_jobs=-1),
        max_iter=10,
        random_state=42,
        verbose=0
    )
    X_imputed = pd.DataFrame(imputer.fit_transform(X_year), columns=X_year.columns)
    
    # Добавляем таргет обратно и сохраняем
    X_imputed["class"] = y_year.values
    X_imputed.to_csv(f"imputed_data/{year}_imputed.csv", index=False)
    
    print(f"{year} сохранён → imputed_data/{year}_imputed.csv")
```
